```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Example assuming you uploaded the file
file_path = '/content/drive/MyDrive/WA_Fn-UseC_-HR-Employee-Attrition.csv'  # Replace with your exact filename if different
df = pd.read_csv(file_path)
```

```python
# Quick look at the dataset
print("Shape:", df.shape)
df.head()
```

```python
# 🧼 Step 2: Data Cleaning & Preprocessing

# Check for missing values
print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
# Drop columns that have no variance or are irrelevant
cols_to_drop = ['EmployeeCount', 'Over18', 'StandardHours']
df = df.drop(columns=cols_to_drop)
```

```python
# Convert target variable "Attrition" to binary (Yes=1, No=0)
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
# Check datatypes and unique counts
print("\nColumn Data Types:")
print(df.dtypes)
print("\nUnique values in each column:")
print(df.nunique())
```

```python
# Optional: check for duplicates
print("\nDuplicate Rows:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
# 📝 Summary statistics
df.describe()
```

```python
#Step 3: Exploratory Data Analysis (EDA)

# Overall Attrition Rate
attrition_rate = df['Attrition'].mean() * 100
print(f"\n📌 Overall Attrition Rate: {attrition_rate:.2f}%")
```

```python
# Attrition by Department
plt.figure(figsize=(8,5))
sns.barplot(x='Department', y='Attrition', data=df, estimator=np.mean)
plt.title('Attrition Rate by Department')
plt.xticks(rotation=20)
plt.show()
```

```python
# Attrition by Job Role
plt.figure(figsize=(10,5))
sns.barplot(x='JobRole', y='Attrition', data=df, estimator=np.mean)
plt.title('Attrition Rate by Job Role')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Monthly Income vs Attrition
plt.figure(figsize=(8,5))
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df)
plt.title('Monthly Income vs Attrition')
plt.show()
```

```python
# Correlation Heatmap for numerical variables
plt.figure(figsize=(12,8))
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()
sns.heatmap(corr, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
#Save cleaned dataset for next steps (modeling & Power BI)
df.to_csv('cleaned_hr_data.csv', index=False)
print("\n✅ Cleaned dataset saved as 'cleaned_hr_data.csv'")
```
